```python
# 5.2 Plot electricity cost for VA, NC, SC
sns.relplot(
    data=df_range[df_range['state'].isin(['Virginia','North Carolina','South Carolina'])],
    x='year',
    y='price_USD',
    hue='state',
    kind='line',
    height=3,
    aspect=4
).set(
    title='Electricity cost: VA, NC, SC',
    xlabel='Year',
    ylabel='Price (USD)'
);
```
