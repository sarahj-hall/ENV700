## 💥New Topic: Formatting output with `print()`💥

The `print()` function can display several values at once by separating them with commas. Python automatically places a space between the values:

````python
site = "Duke Forest"
pH = 6.8

print("Site:", site, "pH:", pH)
````

For more control, use an f-string. Place an f before the opening quotation mark and put variable names inside curly braces:

````python
print(f"Site: {site}; pH: {pH}")
````

F-strings make it easy to combine text and variables in a readable format. They also allow values to be formatted—for example, to display a number with two decimal places:

````python
print(f"Site: {site}; pH: {pH:.2f}")
````

## 💥New topic: Using `enumerate()` in `for` loops💥

When looping through a list, `enumerate()` lets you access both the position and the value of each item:


````python
sites = ["Duke Forest", "Jordan Lake", "Eno River"]

for index, site in enumerate(sites):
    print(index, site)
````

The index starts at 0 by default. To start counting at a different number, use the start argument:

````python
for number, site in enumerate(sites, start=1):
    print(f"{number}. {site}")
````

Using enumerate() is often clearer and safer than manually creating and updating a counter variable.
